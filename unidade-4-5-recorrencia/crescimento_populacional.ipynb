```python
import matplotlib.pyplot as plt
```

## Fórmula de Recorrência Exponencial

**Recorrência:** P(n+1) = P(n) × (1 + r)

**Fórmula Fechada:** P(n) = P₀ × (1 + r)ⁿ

Onde:
- P(n) = população no período n
- r = taxa de crescimento (decimal)
- P₀ = população inicial

```python
def calculadora_crescimento_pop():
  """
    Calcula crescimento populacional usando recorrência exponencial
  """

  try:
    ano_inical = int(input("Digite o ano inicial: "))
    pop_inical = float(input("Digite a população inicial: "))
    taxa_crescimento = float(input("Digite a taxa de crescimento anual(%): "))
    ano_final = int(input("Digite o ano final: "))

    if ano_final <= ano_inical:
      print("\nO ano final deve ser maior que o ano inicial.")
      return


    #converte taxa de porcentagem pra decimal
    taxa_crescimento_decimal = taxa_crescimento / 100

    #cria uma lista com todos os anos
    anos = list(range(ano_inical, ano_final + 1))

    #cria uma lista com a população
    populacoes = [pop_inical]

    #aplica a fórmula de recorrência
    for _ in range(len(anos) - 1):
      nova_pop = populacoes[-1] * (1 + taxa_crescimento_decimal)
      populacoes.append(nova_pop)


    #cria o gráfico
    plt.plot(anos, populacoes)
    plt.title(f'Crescimento Populacional Exponencial\nP(n+1) = P(n) × (1 + {taxa_crescimento}%)', fontsize=12, fontweight='bold')
    plt.xlabel('Ano', fontsize=14)
    plt.ylabel('População', fontsize=14)
    plt.show()

    #exibe resultados finais
    print("\n---- RESULTADOS ----\n")
    print(f"População inicial em {ano_inical}: {pop_inical:,.0f}")
    print(f"Populaçao final em {ano_final}: {populacoes[-1]:,.0f}")
    crescimento = ((populacoes[-1] / pop_inical) - 1) * 100
    print(f"Crescimento total em {len(anos)} anos: {crescimento:.1f}%")
    print(f"Taxa anual: {taxa_crescimento}%")

  except ValueError:
    print("Por favor, digite os valores numéricos válidos.")



```

```python
def main():
  """
  Menu principal do programa
  """
  print("\n---- CALCULADORA DE CRESCIMENTO POPULACIONAL ----")
  calculadora_crescimento_pop()

if __name__ == "__main__":
  main()
```
